### Python code to invoke open VA InterVA and InSilicoVA programs

```python
from pathlib import Path
import subprocess
from dotenv import dotenv_values

directory = dotenv_values(".env").get("OPENVA_PROJECT_DIRECTORY")
if not directory:
    raise ValueError("Set OPENVA_PROJECT_DIRECTORY in the executor's .env file.")

project_directory = Path(directory)
print(f"openVA project directory: {project_directory}")
```

Call InterVA with input/who2016.csv as input and store output in output/who2016-interva5


```python


r_code = """
library(openVA)

data <- read.csv(
    "input/who2016.csv",
    check.names = FALSE,
    stringsAsFactors = FALSE
)

fit <- codeVA(
    data,
    data.type = "WHO2016",
    model = "InterVA",
    version = "5",
    HIV = "h",
    Malaria = "h",
    write = FALSE
)

write.csv(
    getTopCOD(fit, n = 3, include.prob = TRUE),
    "output/who2016-interva5.csv",
    row.names = FALSE
)
"""

result = subprocess.run(
    [
        "docker",
        "compose",
        "run",
        "--rm",
        "openva",
        "Rscript",
        "-e",
        r_code,
    ],
    cwd=project_directory,
    capture_output=True,
    text=True,
    check=True,
)

print(result.stdout)
```

Call InSilicoVA with input/who2016.csv as input and store output in output/who2016-insilico.csv


```python
r_code = """
library(openVA)

data <- read.csv(
    "input/who2016.csv",
    check.names = FALSE,
    stringsAsFactors = FALSE
)

fit <- codeVA(
    data,
    data.type = "WHO2016",
    model = "InSilicoVA",
    Nsim = 10000,
    auto.length = FALSE
)

write.csv(
    getTopCOD(fit, n = 3, include.prob = TRUE),
    "output/who2016-insilico.csv",
    row.names = FALSE
)
"""

result = subprocess.run(
    [
        "docker",
        "compose",
        "run",
        "--rm",
        "openva",
        "Rscript",
        "-e",
        r_code,
    ],
    cwd=project_directory,
    capture_output=True,
    text=True,
    check=True,
)

print(result.stdout)
```
